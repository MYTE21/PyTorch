## ⚙️ Imports

```python
# Libraries: Standard
import logging
import platform

# Libraries: External
import torch
```

## 🌲 Check PyTorch

<div style = "border: 2px solid #4B0082; padding: 5px 10px; border-radius: 10px;">
⚓️ Check the <a href="https://pytorch.org/get-started/locally/">🔗 PyTorch</a> version. ⤵️
</div>

```python
print(f"PyTorch version: {torch.__version__}")
```

<div style = "border: 2px solid #4B0082; padding: 5px 10px; border-radius: 10px;">
⚓️ Check PyTorch has access to <a href="https://pytorch.org/get-started/locally/">🔗 MPS (Metal Performance Shader</a>, Apple's GPU architecture). ⤵️
</div>

```python
def get_mps() -> str | None:
    try:
        print(f"Is MPS (Metal Performance Shader) built?: {torch.backends.mps.is_built()}")
        print(f"Is MPS available?: {torch.backends.mps.is_available()}")

        device_name = "mps" if torch.backends.mps.is_available() else "cpu"
        return device_name
    except Exception as e:
        logging.error(f"MPS is not available: {e}")
```

<div style = "border: 2px solid #4B0082; padding: 5px 10px; border-radius: 10px;">
⚓️ Check PyTorch has access to <a href="https://developer.nvidia.com/cuda-zone">🔗 CUDA</a>, Nvidia (Windows). ⤵️
</div>

```python
def get_cuda() -> str | None:
    try:
        print(f"Is CUDA available?: {torch.cuda.is_available()}")
        print(f"Available GPU devices: {torch.cuda.device_count()}")
        print(f"Current GPU device: {torch.cuda.current_device()}")
        print(f"Current GPU device name: {torch.cuda.get_device_name(torch.cuda.current_device())}")

        device_name = torch.device("cuda" if torch.cuda.is_available() else "cpu")
        return device_name
    except Exception as e:
        logging.error(f"CUDA is not available: {e}")
```

<div style = "border: 2px solid #4B0082; padding: 5px 10px; border-radius: 10px;">
⚓️ Set the device. ⤵️
</div>

```python
device = get_mps() if platform.system() == "Darwin" else get_cuda()
print(f"Using device: {device}")
```

<div style = "border: 2px solid #4B0082; padding: 5px 10px; border-radius: 10px;">
