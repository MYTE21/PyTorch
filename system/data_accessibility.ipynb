## ⚙️ Imports

```python
# Libraries: Standard
import os
import json
import platform
```


## 📁 Verify Dataset Locations

<div style = "border: 2px solid #4B0082; padding: 5px 10px; border-radius: 10px;">
⚓️ Get the <a href="../data/datasets.json">🔗 datasets.json</a> file path. ⤵️
</div>

```python
def get_datasets_file_path() -> str:
    datasets_file_location = "../data/datasets.json"

    with open(datasets_file_location) as file:
        datasets_path = json.load(file)

    return datasets_path
```

<div style = "border: 2px solid #4B0082; padding: 5px 10px; border-radius: 10px;">
⚓️ Get all the <code>dataset</code> paths. ⤵️
</div>

```python
def get_dataset_paths(obj: dict[str] | str):
    values = []

    if isinstance(obj, dict):
        for value in obj.values():
            values.extend(get_dataset_paths(value))
    else:
        values.append(obj)

    return values
```

```python
dataset_paths = get_dataset_paths(get_datasets_file_path())
```

<div style = "border: 2px solid #4B0082; padding: 5px 10px; border-radius: 10px;">
⚓️ Check if all the mentioned datasets are present in the mentioned location. ⤵️
</div>

```python
def files_exists(files: list[str]) -> tuple[int, int]:
    existed_files_count, absent_files_count = 0, 0
    total_files = len(files)

    exit_project_station = files[0] if platform.system() == "Darwin" else files[1]

    for i in range(2, total_files):
        if os.path.exists(exit_project_station + files[i]):
            existed_files_count += 1
        else:
            absent_files_count += 1

    return existed_files_count, absent_files_count
```

```python
file_records = files_exists(dataset_paths)

print(f"1. Existed files count: {file_records[0]}")
print(f"2. Non-Existent files count: {file_records[1]}")
```
